```python
import numpy as np
import matplotlib.pyplot as plt


def log_activation(z):
    return np.log(1+z)


def log_activation_derivative(z):
    return 1/(1+z)


X = np.array([[0],
              [1],
              [2],
              [3],
              [4],
              [5]])

y = np.array([[1],
              [1],
              [1],
              [0],
              [0],
              [0]])


np.random.seed(1)
weight = np.random.randn(1)
bias = np.random.randn(1)

learning_rate = 0.01


for epoch in range(2000):
    linear_output = X * weight + bias
    predictions = log_activation(linear_output)

    error = y - predictions
    d_predictions = error * log_activation_derivative(linear_output)

    weight += learning_rate * np.sum(d_predictions * X)
    bias += learning_rate * np.sum(d_predictions)

print("Training Completed")
print("Final Weight:", weight)
print("Final Bias:", bias)


user_value = float(input("Enter a value to predict: "))


user_array = np.array([[user_value]])


result = log_activation(user_array * weight + bias)

print("Predicted Output:", result)

plt.scatter(X, y, color='red', label='Actual Data')
plt.plot(X, predictions, color='blue', label='Model Prediction')
plt.scatter(user_value, result, color='green', s=100, label='User Input')
plt.xlabel("Input")
plt.ylabel("Output")
plt.title("Neural Network Model with User Input")
plt.legend()
plt.show()
```
